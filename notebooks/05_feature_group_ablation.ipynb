```python
# compare with the paper (Blend test RMSE, m)
pv = res.pivot(index='variant', columns='horizon_d', values='blend')
rows = []
for v, vals in F.PAPER_BLEND.items():
    for h, x in zip((1, 7, 14), vals):
        rows.append((v, h, round(pv.loc[v, h], 4), x))
pd.DataFrame(rows, columns=['variant', 'horizon_d', 'recomputed', 'paper'])
```
